Năm 2010, Haiti hứng chịu trận động đất mạnh 7,0 độ richter. Trận động đất đã gây ra sự tàn phá và thiệt hại tính mạng trên diện rộng và khoảng ba triệu người bị ảnh hưởng bởi thảm họa thiên nhiên này.

````python
haiti = df_can.loc['Haiti', years] # passing in years 1980 - 2013 to exclude the 'total' column
haiti
````

Tiếp theo, chúng ta sẽ vẽ một biểu đồ đường bằng cách thêm `.plot()` vào dataframe `haiti`.

````python
haiti.plot()
````

*pandas* tự động điền trục x với các giá trị chỉ mục (years) và trục y với các giá trị cột (population).

Đặt tên cho các cột x và y bằng cách sử dụng: `plt.title()`, `plt.ylabel()`, và `plt.xlabel()` như sau:

````python
haiti.plot(kind='line')

plt.title('Số lượng nhập cư từ Haiti')
plt.ylabel('Số lượng')
plt.xlabel('Năm')

plt.show()
````

Chúng ta có thể nhận thấy rõ ràng số lượng người nhập cư từ Haiti đã tăng vọt kể từ năm 2010 khi Canada tăng cường nỗ lực tiếp nhận người tị nạn từ Haiti. Hãy chú thích phần tăng đột biến này trong biểu đồ bằng cách sử dụng phương thức `plt.text()`.

````python
haiti.plot(kind='line')

plt.title('Số lượng nhập cư từ Haiti')
plt.ylabel('Số lượng')
plt.xlabel('Năm')

# syntax: plt.text(x, y, label)
plt.text(20, 6000, 'Động đất năm 2010') 

plt.show()
````

Chỉ với một vài dòng mã, bạn có thể nhanh chóng xác định và hình dung sự gia tăng đột biến về số lượng người nhập cư!

Một vài lưu ý về tham số x và y của `plt.text(x, y, label)`:

```
 Bởi vì cột hoành (years) có kiểu dữ liệu là 'integer', nên chúng ta có thể định nghĩa giá trị x là số năm. 
 
 Trục tung y là số lượng người nhập cư cũng có kiểu dữ liệu là 'integer', chúng ta cũng có thể xác định y = 6000.
```

```python
    plt.text(2000, 6000, '2010 Earthquake') # years dưới dạng kiểu dữ liệu integer
```

```
Nếu cột hoành years có kiểu dữ liệu là 'string', Chúng ta cần xác định vị trí cột x ở trong bảng dữ liệu, ví dụ vị trí của năm 2000 là thứ 20 trong bảng dữ liệu.
```

```python
    plt.text(20, 6000, '2010 Earthquake') # years dưới dạng kiểu dữ liệu string
```

*   `barh` for horizontal bar plots
*   `hist` for histogram
*   `box` for boxplot

*   `area` for area plots
*   `pie` for pie plots
*   `scatter` for scatter plots
*   `hexbin` for hexbin plot
