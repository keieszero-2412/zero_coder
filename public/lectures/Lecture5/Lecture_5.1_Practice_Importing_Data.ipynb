```python
# Nhập thư viện pandas (thư viện mạnh mẽ nhất để xử lý dữ liệu bảng trong Python)
# Thường được viết tắt là pd cho ngắn gọn
import pandas as pd
```

## Đọc dữ liệu từ file CSV 

```python
# Load file csv từ thư mục máy tính

file_source = "datasets/automotive_data/data.csv"

csv_df = pd.read_csv(file_source)
```

```python
csv_df.head()
```

```python
# Đọc file CSV bằng Python – không có tiêu đề 
csv_df = pd.read_csv(file_source, header=None)
```

```python
csv_df
```

```python
# Lấy 10 dòng đầu tiên
csv_df_10_head = csv_df.head(10)
```

```python
csv_df_10_head
```

```python
# Lấy 10 dòng cuối cùng
csv_df_10_tail = csv_df.tail(10)
```

```python
csv_df_10_tail
```

```python
# Thêm tiêu đề
headers = ["symboling","normalized-losses","make","fuel-type","aspiration", "num-of-doors","body-style",
         "drive-wheels","engine-location","wheel-base", "length","width","height","curb-weight","engine-type",
         "num-of-cylinders", "engine-size","fuel-system","bore","stroke","compression-ratio","horsepower",
         "peak-rpm","city-mpg","highway-mpg","price"]

csv_df.columns = headers
csv_df.head(10)
```

```python
# Xuất dataframe ra file CSV
csv_df.to_csv('./datasets/automotive_data/data_with_header.csv')
```

# Load file json từ thư mục máy tính


JSON là viết tắt của JavaScript Object Notation, là một kiểu định dạng dữ liệu tuân theo một quy luật nhất định mà hầu hết các ngôn ngữ lập trình hiện nay đều có thể đọc được. JSON là một tiêu chuẩn mở để trao đổi dữ liệu trên web.

Định dạng JSON sử dụng các cặp key – value để dữ liệu sử dụng. Nó hỗ trợ các cấu trúc dữ liệu như đối tượng và mảng. 

{
"employees":[
    {"firstName":"John", "lastName":"Doe"},
    {"firstName":"Anna", "lastName":"Smith"},
    {"firstName":"Peter", "lastName":"Jones"}
]
}

```python
# Load file json từ thư mục máy tính
json_file = "./datasets/batdongsan.json"
```

```python
json_df = pd.read_json(json_file)
```

```python
json_df.head()
```

```python
json_df_10_tail = json_df.tail(10)
json_df_10_tail
```

```python
# Xuất dataframe ra file json
json_df_10_tail.to_json("./datasets/bds_10_tail.json")
```
