# Làm việc và tương tác với cơ sở dữ liệu

## Kết nối với cơ sở dữ liệu

### Import thư viện có thể tương tác với cơ sở dữ liệu

Trong ví dụ này chúng ta sử dụng thư viện pymssql để kết nối với cơ sở dữ liệu MSSQL. Đối với các cơ sở dữ liệu khác như Mysql, Postgres, Netteza sẽ có những thư viện python khác hỗ trợ để phục vụ việc kết nối tới các cơ sở dữ liệu.

[Tìm hiểu thêm về thư viện pymssql tại đây!](https://pymssql.readthedocs.io/en/stable/intro.html).



```python
# Import thư viện cần thiết
import pymssql
import pandas as pd
```

### Khởi tạo một kết nối tới cơ sở dữ liệu sử dụng hàm connect trong thư viện pymssql

```python
# Khai báo thông tin server name, user, password, database name để kết nối với server mssql
conn = pymssql.connect(server="sql-st.datapot.edu.vn", user="datapot_student", password="datapot@mspartner", database="adventureworks")
cursor = conn.cursor()
```

```python
# Chạy câu lệnh SQL để truy vấn dữ liệu
cursor.execute ("SELECT * from [adventureworks].[SalesLT].[Customer]")
```

```python
# Kiểm tra các tên cột của bảng customer trong db adventurework
num_fields = len(cursor.description)
field_names = [i[0] for i in cursor.description]
print(field_names)
```

```python
# Lấy toàn bộ dữ liệu trong database
row = cursor.fetchall()
print("Mẫu dữ liệu lấy được: \n",row[1])
print("Đếm kích thước của bảng customer: ", len(row))
```

```python
# Tạo một pandas dataframe từ kết quả và tên cột đã lấy được
df = pd.DataFrame(row, columns=field_names)
df.head()
```

```python
# Sau khi truy vấn thành công, tắt session kết nối sau khi không có nhu cầu sử dụng
cursor.close()
conn.close()
```

```python
# Tổng hợp toàn bộ các bước nhỏ thành một luồng tổng thể:
import pymssql
import pandas as pd
try:
    conn = pymssql.connect(server="sql-st.datapot.edu.vn", user="datapot_student", password="datapot@mspartner", database="adventureworks")
    cursor = conn.cursor()
    cursor.execute ("SELECT * from [adventureworks].[SalesLT].[Customer]")
    num_fields = len(cursor.description)
    field_names = [i[0] for i in cursor.description]
    row = cursor.fetchall()
    df = pd.DataFrame(row, columns=field_names)
    cursor.close()
    conn.close()
except Exception:
    print("\nERROR: Unable to connect to the server.")
    exit(-1)
    
df.head()
```
