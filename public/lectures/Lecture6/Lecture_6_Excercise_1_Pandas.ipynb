# Cấu trúc dữ liệu <a name="structures"></a>

## Dữ liệu dạng mảng (Series) <a name="series"></a>

`Series` là một đối tượng giống như mảng một chiều chứa một chuỗi các giá trị và một mảng nhãn dữ liệu liên quan được gọi là chỉ mục của nó.

Dưới đây là cách tạo một mảng bằng pandas

````python
data = pd.Series([4, 7, -5, 3])
````

Print mảng vừa được tạo

````python
data
````

Các chỉ mục của mảng nằm ở bên trái và các giá trị ở bên phải. 

Nếu không định nghĩa chỉ mục cho từng mảng, thì chỉ mục mặc định của mảng là các số nguyên từ 0 đến N-1.

Chúng ta có thể lấy các giá trị của mảng bảng cách:

````python
data.values
````

Chúng ta có thể chỉ định các chỉ mục tùy chỉnh khi khởi tạo một mảng

````python
data2 = pd.Series([4, 7, -5, 3], index=["a", "b", "c", "d"])
````

````python
data2
````

Chúng ta có thể tạo một mảng từ một dictionary

````python
cities = {"Glasgow" : 599650, "Edinburgh" : 464990, "Aberdeen" : 196670, "Dundee" : 147710}
data3 = pd.Series(cities)
````

````python
data3
````

## DataFrame <a name="dataframe"></a>

Dataframe biểu thị một bảng dữ liệu hình chữ nhật và chứa một tập hợp các cột được sắp xếp theo thứ tự, mỗi cột có thể là một loại giá trị khác nhau. DataFrame có cả chỉ mục hàng và cột.

Cách đơn giản nhất để tạo ra một dataframe là từ một dictionary:

````python
data = {"cities" : ["Glasgow", "Edinburgh", "Aberdeen", "Dundee"],
        "population" : [599650, 464990, 196670, 147710],
        "year" : [2011, 2013, 2013, 2013]}
df = pd.DataFrame(data)
````

kiểm tra df vừa mới được tạo:

````python
df.cities #call column
df.iloc[0:3:2] #call row 1 and row 3
````

Bạn có thể sắp xếp thứ tự các cột, chỉ mục của các hàng trong quá trình tạo một dataframe:

````python
df2 = pd.DataFrame(data, columns=["year", "cities", "population"], index=["a", "b", "c", "d"])
df2
````

Bạn có thể chọn 1 cột trong dataframe bằng cách:
```python
df2["cities"]
```
Kết quả trả về sẽ là một mảng trong python:

````python
df2["cities"]
````

It is also possible to add and modify the columns of a DataFrame

Bạn cũng có thể thêm và sửa dữ liệu các cột trong python

````python
df2["size"] = 100
df2
````

````python
df2["size"] = [175, 264, 65.1, 60]  # in km^2
df2
````

# Một số hàm quan trọng <a name="ess_func"></a>

Trong phần này, chúng ta sẽ tìm hiểu các cơ chế cơ bản để tương tác với dữ liệu dạng mảng hoặc DaraFrame.

## Xoá cột hoặc chỉ mục <a name="removing"></a>

Chúng ta có thể sử dụng hàm `drop()` để xoá dữ liệu, nó có thể áp dụng được với cột và các chỉ mục:

````python
# Tạo một dataframe mới:
data = np.reshape(np.arange(9), (3,3))
df = pd.DataFrame(data, index=["a", "b", "c"],
                  columns=["Edinburgh", "Glasgow", "Aberdeen"])

df
````

````python
df.drop("b")  # xoá hàng dựa vàoc chỉ mục
````

````python
# Bạn cũng có thể xoá các cột (axis=1 or axis="columns")
df.drop(["Aberdeen", "Edinburgh"], axis="columns")
````

Lưu ý rằng dataframe gốc không thay đổi nếu chúng ta sử dụng: `df.drop()`. Bởi vì khi sử dụng `df.drop()`, nó giữ nguyên dữ liệu gốc và tạo ramột dataframe mới với dữ liệu mong muốn đã bị loại bỏ. 

Để áp dụng `.drop()` vào trực tiếp dataframe gốc, chúng ta có thể thực hiện bằng cách thêm argument: `inplace=True` vào hàm.

````python
df
````

## Indexing <a name="indexing"></a>

Chúng ta sử dụng `.iloc` để chỉ mục dữ liệu trong dataframe

````python
data = np.reshape(np.arange(9), (3,3))
df = pd.DataFrame(data, index=["a", "b", "c"],
                  columns=["Edinburgh", "Glasgow", "Aberdeen"])

df
````

````python
# Thay vì df[2] (bị lỗi KeyError do tìm theo tên cột), dùng df.iloc[2] để lấy dòng theo vị trí:
df.iloc[2]
````

````python
df.iloc[2]  # this works and return a series
````

Bạn cũng có thể cắt một dataframe như bình thường:

````python
df.iloc[:2]
````

````python
df.iloc[2,0]
````

Khi các chỉ mục/tên cột của dataframe của bạn không phải là số nguyên, bạn có thể sử dụng `.loc` để lập chỉ mục:

````python
df2
````

````python
df2.loc["a", "size"]
````

## Tóm tắt và tính toán số liệu thống kê mô tả<a name="sums"></a>
`pandas` được trang bị các phương pháp toán học và thống kê phổ biến:

Tạo một dataframe mới:

````python
df = pd.DataFrame(np.arange(20).reshape(5,4),
                 columns=["a", "b", "c", "d"])
df
````

Tính tổng giá trị theo chiều dọc:

````python
df.sum()
````

Tính tổng giá trị theo chiều ngang:

````python
df.sum(axis="columns")
````

Tính giá trị trung bình theo chiều dọc:

````python
df.mean()
````

Tính tổng giá trị theo chiều ngang:

````python
df.mean(axis="columns")
````

Chúng ta có thể tính toán các bước trên bàng 1 hàng duy nhất, hàm: `describe()` 

Hàm `describe()`  sẽ tính toán các thông số thống kê mô tả của dataframe:

````python
df.describe()
````

Đây là một vài phương thức chính thường được sử dụng với Dataframe:

| Phương thức | Mô tả |
| -- | -- |
| count          | Trả ra kết quả số lượng giá trị khác null |
| describe       | Trả ra kết quả thống kê mô tả của dataframe |
| min, max       | Trả ra kết quả thấp nhất, cao nhất của dataframe |
| argmin, argmax | Trả ra chỉ mục của giá trị thấp nhất, cao nhất trong dataframe | 
| sum            | Trả ra kết quả tính tổng |
| mean           | Trả ra kết quả tính trung bình |
| median         | Trả ra kết quả tính trung vị |
| std            | Trả ra kết quả Độ lệch chuẩn
| value_counts() | Đếm số lần xuất hiện của từng phần tử duy nhất trong một cột |
