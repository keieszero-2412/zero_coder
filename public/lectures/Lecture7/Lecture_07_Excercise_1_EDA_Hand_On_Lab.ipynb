<h2 id="basic_grouping">4. Khái niệm cơ bản về phân nhóm dữ liệu</h2>


Phương pháp "groupby" nhóm dữ liệu theo các danh mục khác nhau. Dữ liệu được nhóm dựa trên một hoặc một số biến và phân tích được thực hiện trên từng nhóm riêng lẻ.

Ví dụ: hãy nhóm theo biến "drive-wheels". Chúng ta thấy rằng có 3 loại bánh dẫn động khác nhau.

```python
df['drive-wheels'].unique()
```

<p>Nếu muốn biết trung bình loại bánh dẫn động nào có giá trị nhất, chúng ta có thể nhóm các "drive-wheels" rồi tính trung bình cho biến này.</p>

<p>Chúng ta có thể gán một dataframe mới có 3 cột: drive-wheels', 'body-style' và 'price' để phục vụ phân tích.</p>



```python
df_group_one = df[['drive-wheels','body-style','price']]
```

Chúng ta có thể tính giá trung bình cho từng loại dữ liệu khác nhau:

```python
# grouping results
df_group_one = df_group_one.groupby(['drive-wheels'],as_index=False).mean(numeric_only=True)
df_group_one
```

Từ dữ liệu cho ta thấy có vẻ như xe  4wd (Four Wheel Drive) nhìn chung là đắt nhất, trong khi xe fwd và rwd có giá xấp xỉ nhau.

Học viên cũng có thể nhóm với nhiều biến. Ví dụ: hãy nhóm theo cả 'drive-wheels' và 'body-style'. 

Học viên hãy lưu trữ kết quả trong dataframe mới có tên 'grouped_test1'

```python
# grouping results
df_gptest = df[['drive-wheels','body-style','price']]
grouped_test1 = df_gptest.groupby(['drive-wheels','body-style'],as_index=False).mean()
grouped_test1
```

Bảng tổng hợp (pivot table) sẽ giúp chúng ta hiểu hơn nhiều về bảng dữ liệu kiểu như thế này.

Bảng tổng hợp giống như một bảng tính Excel, với một biến dọc theo cột và một biến khác dọc theo hàng. Chúng ta có thể chuyển đổi khung dữ liệu thành bảng tổng hợp bằng phương pháp "pivot" để tạo bảng tổng hợp từ các nhóm.


<p>Trong trường hợp này, chúng ta sẽ để biến bánh xe dẫn động (drive-wheels) làm các hàng của bảng và kiểu thân trục xoay (body-style) sẽ trở thành các cột của bảng:</p>


```python
grouped_pivot = grouped_test1.pivot(index='drive-wheels',columns='body-style')
grouped_pivot
```

<p>Thông thường, chúng ta sẽ không có dữ liệu cho một số ô tổng hợp. Chúng ta có thể điền giá trị 0 vào các ô bị thiếu này, tuy nhiên bất kỳ giá trị nào khác cũng có thể được sử dụng. 
    
[Tham khảo thêm về phần xử lý dữ liệu bị thiếu]</p>

```python
grouped_pivot = grouped_pivot.fillna(0) # Thay thế dữ liệu còn thiếu bằng giá trị 0
grouped_pivot
```

<h2 id="correlation_causation">5. Mối quan hệ tương quan (Correlation) và nhân quả (Causation:)</h2>


<p><b>Correlation - Tương quan</b>: Thước đo mức độ phụ thuộc lẫn nhau giữa các biến.</p>

<p><b>Causation - Nhân quả</b>: Mối quan hệ giữa nguyên nhân và kết quả giữa hai biến.</p>

<p> Một mối quan hệ tương quan giữa 2 biến không hàm ý 2 biến đó có mối quan hệ nhân quả. 
    Việc xác định mối tương quan đơn giản hơn nhiều so với việc xác định quan hệ nhân quả vì quan hệ nhân quả có thể yêu cầu các phép thử nghiệm độc lập.</p>


<p3>Pearson Correlation (Tương quan Pearson)</p>

<p>Tương quan Pearson đo lường sự phụ thuộc tuyến tính giữa hai biến X và Y.</p>
<p>Hệ số tương quan Pearson là giá trị bao gồm từ -1 đến 1, trong đó:</p>
<ul>
    <li><b>1</b>: Tương quan tuyến tính dương tuyệt đối.</li>
    <li><b>0</b>: Không có mối tương quan tuyến tính, hai biến này không ảnh hưởng lẫn nhau.</li>
    <li><b>-1</b>: Tương quan tuyến tính âm tuyệt đối.</li>
</ul>


Pearson Correlation là phương pháp mặc định được sử dụng trong hàm `corr` của thư viện `pandas` để tính mối tương quan giữa các biến. Chúng ta có thể sử dụng `corr` đểtính toán Tương quan Pearson của các biến 'int64' hoặc 'float64'. 


```python
df.corr(numeric_only=True)
```

Đôi khi chúng ta muốn biết ý nghĩa của ước lượng tương quan có mang ý nghĩa thống kê hay không? Chúng ta có kiểm định thống kê và sử dụng hệ số `P-value`

<b>P-value</b>: 

Giá trị P-value là gì?

P-value là viết tắt của từ probability value, đây là một con số xác suất và được gọi là trị số P. Hiểu một cách đơn giản, đây là một đại lượng giúp các nhà khoa học hay các chuyên gia quyết định giả thuyết của họ đúng hay sai. Thông thường, P value được xác định ở mức ý nghĩa là 0,05, có nghĩa là 95% mực độ tin cậy rằng mối tương quan giữa các biến là đáng kể.

Theo quy ước, khi

<ul>
    <li>p-value &lt; 0.001: Chúng ta có thể khẳng định rằng có bằng chứng mạnh cho thấy mối tương quan là có ý nghĩa về mặt thống kê.</li>
    <li>p-value &lt; 0.05: Chúng ta có thể khẳng định rằng có bằng chứng vừa phải cho thấy mối tương quan là có ý nghĩa về mặt thống kê.</li>
    <li>p-value &lt; 0.1: Chúng ta có thể khẳng định rằng có bằng chứng yếu cho thấy mối tương quan là có ý nghĩa về mặt thống kê.</li>
    <li>p-value &gt; 0.1: Chúng ta có thể khẳng định rằng không có bằng chứng cho thấy mối tương quan là có ý nghĩa về mặt thống kê.</li>
</ul>


Chúng ta có thể xác định được trị số P value bằng các sử dụng module "stats" trong thư viện "scipy"

```python
from scipy import stats
```

<h3>Wheel-base vs Price</h3>


Cùng tính toán hệ số tương quan Pearson và trị số P-Value của 2 biến 'wheel-base' và 'price'. 

```python
pearson_coef, p_value = stats.pearsonr(df['wheel-base'], df['price'])
print("Hệ số tương quan Pearson là", pearson_coef, "với giá trị P-value =", p_value)  
```

    mặc dù mối quan hệ tuyến tính không thực sự là mạnh (~0.585)</p>

<h2>6. Extra: ANOVA</h2>


<h3>ANOVA: Phân tích phương sai</h3>

Kiểm định ANOVA (ANOVA test), hay còn được gọi là phân tích phương sai (Analysis of Variance), là một kỹ thuật thống kê tham số được sử dụng để phân tích sự khác nhau giữa giá trị trung bình của các biến phụ thuộc với nhau. Kết qủa phân tích phương sai trả về 2 tham số:

<p><b>F-test score</b>: F-test score là một thống kê thường được sử dụng trong phân tích phương sai (ANOVA) để kiểm tra sự khác biệt giữa các giá trị trung bình của hai hay nhiều nhóm dữ liệu. Nếu tỷ lệ F cao, điều đó có nghĩa là giá trị trung bình của hai nhóm khác nhau đáng kể. Nếu giá trị F nhỏ, điều đó có nghĩa là không có sự khác biệt đáng kể giữa giá trị trung bình của hai nhóm.</p>

<p><b>P-value</b>:  P-value chỉ ra mức độ ý nghĩa về mặt thống kê mà chỉ số chúng ta đã tính toán ra.</p>

<p>Nếu biến price của chúng ta có mối tương quan chặt chẽ với biến chúng ta đang phân tích, Kiểm định ANOVA sẽ trả về điểm kiểm tra F khá lớn và giá trị p nhỏ.</p>

Chúng ta có thể lấy được các giá trị của hàm `group` bằng các sử dụng hàm `get_group`

```python
grouped_test2.get_group('4wd')['price']
```

Chúng ta có thể sử dụng hàm `f_oneway` trong module `stats` để xác định các chỉ số <b>F-test score</b> và <b>P-value</b>.

```python
# ANOVA
f_val, p_val = stats.f_oneway(grouped_test2.get_group('fwd')['price'], grouped_test2.get_group('rwd')['price'], grouped_test2.get_group('4wd')['price'])  
 
print( "ANOVA results: F=", f_val, ", P =", p_val)   
```

Đây là một kết quả tuyệt vời, với điểm kiểm định F_score lớn cho thấy mối tương quan chặt chẽ và giá trị P gần như bằng 0 cho thấy ý nghĩa thống kê gần như chắc chắn. Nhưng điều này có nghĩa là cả ba nhóm được thử nghiệm đều có mối tương quan cao như vậy?

#### Separately: fwd and rwd

#### 4wd and rwd
