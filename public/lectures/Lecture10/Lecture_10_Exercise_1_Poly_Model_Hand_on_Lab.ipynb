<b>Hôi quy đa thức là gì?</b> Là một trường hợp đặc biệt trong các bài toán hồi quy:
Được sử dụng để mô tả các điểm phân bố, phân tán dưới dạng đường cong

Sử dụng mối quan hệ giữa các biến độc lập x và biến phụ thuộc y được biểu diễn dưới dạng đa thức bậc n và tìm cách tốt nhất vẽ một đường qua các điểm dữ liệu sao cho tối ưu và phù hợp nhất.

Sau đây là một số ví dụ về hồi quy đa thức.

<center><b>Quadratic - Đa thức bậc 2</b></center>

$$
\hat{Y} = a + b_1 X + b_2 X^2
$$

<center><b>Cubic - Đa thức bậc 3</b></center>

$$
\hat{Y} = a + b_1 X + b_2 X^2 + b_3 X^3
$$

<center><b>Đa thức bậc cao hơn:</b></center>

$$
Y = a + b_1 X + b_2 X^2 + b_3 X^3 + \dots
$$


Biểu thức cho hàm đa thức đa biến trở nên phức tạp hơn rất nhiều so với hồi quy đa thức một biến thông thường. Ví dụ: biểu thức của đa thức bậc hai, có hai biến được cho bởi:

$$
Yhat = a + b_1 X_1 +b_2 X_2 +b_3 X_1 X_2+b_4 X_1^2+b_5 X_2^2
$$

<h2>Pipeline</h2>


<p>Data Pipelines đơn giản hóa các bước xử lý dữ liệu. Trong phần này chúng ta sử dụng module <b>Pipeline</b> để tạo Pipeline. Chúng ta cũng sử dụng <b>StandardScaler</b> làm một bước trong quy trình của mình.</p>

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

Chúng ta tạo pipeline bằng cách tạo danh sách các tuples dữ liệu bao gồm tên của mô hình hoặc công cụ ước tính và hàm tương ứng của nó.

```python
Input=[('scale',StandardScaler()), 
       ('polynomial', PolynomialFeatures(include_bias=False)), 
       ('model',LinearRegression())]
```

Đưa danh sách các bước vừa tạo ở trên làm input đầu vào cho `Pipeline`

```python
pipe=Pipeline(Input)
pipe
```

Chúng ta có thể chuẩn hóa dữ liệu, thực hiện chuyển đổi và điều chỉnh mô hình cùng một lúc

```python
pipe.fit(Z,Y)
```

Tương tự, chúng ta có thể chuẩn hóa dữ liệu, thực hiện chuyển đổi và đưa ra dự đoán đồng thời

```python
ypipe=pipe.predict(Z)
ypipe[0:5]
```

<h2>Phần 2: Đánh giá mức độ hiệu quả của mô hình</h2>


<p>Khi đánh giá các mô hình của mình, chúng ta không chỉ muốn trực quan hóa kết quả mà còn muốn có một thước đo định lượng để xác định mức độ chính xác của mô hình.</p>

<p>2 chỉ số đo lường quan trọng để xác định sự phù hợp của một mô hình:</p>
<ul>
    <li><b>Mean Squared Error (MSE)</b></li>
    <li><b>R^2 / R-squared</b></li>
</ul>
    

<b>Mean Squared Error (MSE)</b>

<p>Sai số bình phương trung bình đo mức trung bình của các bình phương sai số, nghĩa là chênh lệch giữa giá trị thực tế (y) và giá trị ước tính (ŷ).</p>

<b>R-squared</b>
<p>R-squared cho ta biết mức độ các biến đầu vào sẽ giải thích được bao nhiêu phần trăm các biến mục tiêu</p>
    
<p>R-squared càng lớn thì mô hình càng tốt. Khi R-squared bằng 95% điều đó có nghĩa rằng các biến đầu vào đã giải thích được 95% sự biến động của biến mục tiêu</p>

<h3>Mô hình 1: Mô hình Linear Regression</h3>


```python
X = df[['horsepower']]
```

Tính toán chỉ số R^2


```python
# horsepower_fit
lm = LinearRegression()
lm.fit(X, Y)
# Find the R^2
print('Chỉ số R-square = ', lm.score(X, Y))
```

Chỉ số R-square =  0.6554631732894476 điều đó có nghĩa rằng các biến đầu vào đã giải thích được ~ 65.546% sự biến động của biến mục tiêu

Tính toán chỉ số MSE

Chúng ta có thể dự đoán đầu ra, tức là "yhat" bằng phương thức `.predict(X)`, trong đó X là biến đầu vào:

```python
Yhat=lm.predict(X)
print('The output of the first four predicted value is: ', Yhat[0:5])
```

Bây giờ hãy import <b>mean_squared_error</b> từ module <b>metrics</b>


```python
from sklearn.metrics import mean_squared_error
```

Chúng ta hãy so sánh kết quả dự đoán với kết quả thực tế

```python
mse = mean_squared_error(Y, Yhat)
print('MSE giữa kết quả thực tế và kết quả dự đoán là: ', mse)
```
